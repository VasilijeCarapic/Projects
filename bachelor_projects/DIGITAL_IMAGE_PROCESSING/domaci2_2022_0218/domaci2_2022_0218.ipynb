# Diskretna Furijeova transformacija 2D Gausove funkcije

Dvodimenzionalna Gausova funkcija dobija se tako što pomnožimo jednodimenzionu gausovu funkciju po kolonama:

$$
g(c) = e^{-\frac{c^2}{2\sigma_{cS}^2}}
$$

i jednodimenzionu gausovu funkciju po redovima: 

$$
g(r) = e^{-\frac{r^2}{2\sigma_{rS}^2}}
$$

nakon čega dobijamo dvodimenzionalnu gausovu funkciju:

$$
g(r,c) = g(r)g(c) = e^{-\frac{r^2}{2\sigma_{rS}^2} - \frac{c^2}{2\sigma_{cS}^2}}
$$
gde su:
- Širina i visina slike: \(NW\) i \(NH\)  
- Broj tačaka DFT po horizontali i vertikali: \(NW\) i \(NH\)  

---
Da bi dokazali da je i u frekvencijskom domenu isto tako gausova funkcija kao i u prostornom domenu potrebno je izračunati DFT po kolonama i DFT po redovima kako bi to pomnožili i dobili rezultat u frekv. domenu.  
Ako je Gausova funkcija i u frekv domenu nakon primene DFT-a na dvodimenzionalnu funkciju dokazali smo ono što je potrebno.

## 1. DFT po redovima i kolonama

Formule za DFT po redovima kao i po kolonama date su ispod.  
1D diskretna Furijeova transformacija po redovima:

$$
G_r(k) = \sum_{r=0}^{N-1} g_r(r) \, e^{-j 2\pi \frac{k}{N_H} r}
$$

1D diskretna Furijeova transformacija po kolonama:

$$
\
G_c(l) = \sum_{c=0}^{NW-1} g_c(c) \, e^{-j 2\pi \frac{l}{N_w} c}
\
$$ 

---

## 2. Aproksimacija integralom

Pošto Gauss funkcija brzo opada, sume od 0 do N-1 možemo aproksimirati kao sume od -∞ do +∞, odakle se dobija:

$$
G_r(k) = \sum_{r=-\infty}^{+\infty} g_r(r) \, e^{-j 2\pi \frac{k}{N_H} r}
$$

i

$$
G_c(l) = \sum_{c=-\infty}^{+\infty} g_c(c) \, e^{-j 2\pi \frac{k}{N_W} l}
$$

A nakon toga možemo tu sumu aproksimirati integralom koji ako malo bolje pogledamo jeste Furijeova transformacija, pa rešavanjem integrala dobijemo sledeći rezultat:  

## Gr(k) = FT{g(r)}:
$$
G_r(k) \approx \int_{-\infty}^{\infty} e^{-r^2/(2\sigma_{rS}^2)} e^{-j 2\pi \frac{k}{N_H} r} dc
= \sigma_{rS} \sqrt{2\pi} \, e^{-2 \pi^2 \sigma_{rS}^2 \frac{k^2}{N_H^2}}
$$
 
## Gc(l) = FT{g(c)}: 

$$
G_c(l) \approx \int_{-\infty}^{\infty} e^{-c^2/(2\sigma_{cS}^2)} e^{-j 2\pi \frac{l}{N_W} c} dc
= \sigma_{cS} \sqrt{2\pi} \, e^{-2 \pi^2 \sigma_{cS}^2 \frac{l^2}{N_W^2}}
$$

Čime se dobijaju furijeove transformacije kao. Gledajući definiciju  

---

## 3. 2D DFT

2D diskretna Furijeova transformacija se dobija kao proizvod 1D transformacija po redovima i kolonama:

$$
G_F(k,l) = G_r(k) \cdot G_c(l)
= C e^{-2 \pi^2 \sigma_{rS}^2 \frac{k^2}{N_H^2} - 2 \pi^2 \sigma_{cS}^2 \frac{l^2}{N_W^2}}
$$

gde je:
$$
C = \sigma_{rS} \sigma_{cS} 2\pi 
$$
konstanta normalizacije.

## 4. Standardne devijacije u frekvencijskom domenu
- 2D DFT Gaussove funkcije je **takođe Gausova** sto se lako da proveriti jer obe imaju isti oblik Gausove funkcije.  
- Ako uporedimo sa recimo ovom funkcijom u prostornom domenu $$
g(r,c) = g(r)g(c) = e^{-\frac{r^2}{2\sigma_{rS}^2} - \frac{c^2}{2\sigma_{cS}^2}}
$$ mozemo lako odrediti parametar $$\sigma_{rS}$$ i parametar $$\sigma_{cS}$$ 

- Standardne devijacije u frekvencijskom domenu zavise od standardnih devijacija u prostornom domenu i broja tačaka diskretizacije:  
$$
\sigma_{rF}^2 = f(\sigma_{rS}, N_H) = \frac{N_H^2}{4*\pi^2*\sigma_{rS}^2}, \quad \sigma_{cF}^2 = f(\sigma_{cS}, N_W) = \frac{N_W^2}{4*\pi^2*\sigma_{cS}^2}
$$

---
Diskretna DFT Gauss funkcija je **takođe Gaussova**, sa standardnim devijacijama:
$$
\sigma_r^F = \frac{N_H}{2\pi \sigma_{rS}}, \quad
\sigma_c^F = \frac{N_W}{2\pi \sigma_{cS}}
$$

- Odnos standardnih devijacija u prostornom i frekvencijskom domenu:

$$
\sigma_F \sim \frac{N}{2\pi \sigma^S}
$$

````python
def lpfilter(filt_type, Ny, Nx, sigma, n=1):
    
    if (Ny%2 == 0):
        y = np.arange(0,Ny) - Ny/2 + 0.5
    else:
        y = np.arange(0,Ny) - (Ny-1)/2
    
    if (Nx%2 == 0):
        x = np.arange(0,Nx) - Nx/2 + 0.5
    else:
        x = np.arange(0,Nx) - (Nx-1)/2

    
    X, Y = np.meshgrid(x, y)
    
    D = np.sqrt(np.square(X) + np.square(Y))
    
    if filt_type == 'gaussian':
        filter_mask = np.exp(-np.square(D)/(2*np.square(sigma)))
    elif filt_type == 'btw':
        filter_mask = 1/(1+(D/sigma)**(2*n))
    elif filt_type == 'ideal':
        filter_mask = np.ones([Ny, Nx])
        filter_mask[D>sigma] = 0
    else:
        print('Greška! Nije podržan tip filtra: ', filt_type)
        return
    
    return filter_mask
````

Funkcija `filter_image` primenjuje **niskopropusno filtriranje slike u frekvencijskom domenu**.  
Ulazna slika se najpre transformiše u frekvencijski domen pomoću dvodimenzionalne Furijeove transformacije (FFT), čime se dobija njena spektralna reprezentacija.  

Zatim se generiše Gaussian niskopropusni filter, koji potiskuje visoke frekvencije i omogućava zaglađivanje slike.  
Filtriranje se ostvaruje množenjem spektra slike i frekvencijskog filtra.  

Nakon filtriranja, slika se vraća u prostorni domen primenom inverzne Furijeove transformacije (IFFT).  
Na kraju se vrednosti piksela ograničavaju na opseg \([0, 1]\), čime se obezbeđuje ispravan prikaz filtrirane slike.


````python
def filter_image(img_org, sigma):
    [Ny, Nx] = np.shape(img_org)

    img = img_org.copy()
    #slika u frekv domenu
    img_fft = np.fft.fftshift(np.fft.fft2(img))

    #filtar
    filt_freq = lpfilter("gaussian", Ny, Nx, sigma)

    #filtrirana slika

    img_fft_filt = img_fft * filt_freq

    # vrati nazad u realni domen
    img_filt = np.real(np.fft.ifft2(np.fft.ifftshift(img_fft_filt)))
    img_filt = np.clip(img_filt, 0, 1)

    return img_filt
````

````python
def plot_img(img, filtered_img):
    '''prikazuje slika, filtrirana slika na jednom grafiku '''
    fig, axes = plt.subplots(1, 2, figsize = (10, 6), dpi = 100)
    ax = axes.ravel()

    ax[0].imshow(img, cmap = "gray"); ax[0].set_axis_off();ax[0].set_title("Ulazna slika");
    ax[1].imshow(filtered_img, cmap = "gray"); ax[1].set_axis_off();ax[1].set_title("Filtrirana slika");

    plt.tight_layout()
    plt.show()
````

## 1. c)

Funkcija `lpfilter_anisotrop` generiše **anizotropni niskopropusni filter u frekvencijskom domenu**.  
Filter se formira za zadate dimenzije slike (`Ny`, `Nx`) i različite standardne devijacije po x i y osi (`sigma_x`, `sigma_y`), čime se omogućava anizotropno filtriranje.

Najpre se formiraju koordinate u frekvencijskom domenu, pri čemu se vodi računa o parnim i neparnim dimenzijama slike kako bi se frekvencijska osa pravilno centrirala.  
Zatim se izračunava normalizovano rastojanje `D`, koje uzima u obzir različite vrednosti `sigma_x` i `sigma_y`.

U zavisnosti od parametra `filt_type`, funkcija može generisati Gaussian, Butterworth (`btw`) ili idealni niskopropusni filter.  
Na kraju se vraća maska filtra koja se koristi za frekvencijsko filtriranje slike.


````python
def lpfilter_anisotrop(filt_type, Ny, Nx, sigma_x, sigma_y, n = 1):
    if(Ny % 2 == 0):
        # 0 1 2 3 => (-2 -1 0 1) + 0.5 => (-1.5 -0.5 0.5 1.5)
        y = np.arange(0, Ny) - (Ny / 2) + 0.5
    else:
        y = np.arange(0, Ny) - (Ny - 1) / 2

    if(Nx % 2 == 0):
        x = np.arange(0, Nx) - (Nx / 2) + 0.5
    else:
        x = np.arange(0, Nx) - (Nx - 1) / 2

    X, Y = np.meshgrid(x, y)

    #izracunam D kao
    D = np.sqrt(np.square(X / sigma_x) + np.square(Y / sigma_y))

    if filt_type == "gaussian":
        filter_mask = np.exp(-X**2 / (2 * sigma_x ** 2) - Y ** 2 / (2 * sigma_y ** 2))
    elif filt_type == "btw":
        filter_mask = 1 / (1 + D **(2 * n))
    elif filt_type == "ideal":
        filter_mask = np.ones([Ny, Nx])
        filter_mask[D > 1] = 0
    else:
        print("Error! That filter type is not available", filt_type)
        return
    return filter_mask
````

S obzirom na to da se zahteva da parametri filtra u frekvencijskom domenu budu ekvivalentni parametrima u prostornom domenu, kao i da je filter **izotropan**  
($\sigma_{s_x} = \sigma_{s_y} = \sigma_s$), moguće je odrediti odgovarajuće parametre u frekvencijskom domenu.

Standardne devijacije u frekvencijskom domenu računaju se na sledeći način:

$$
\sigma_{f_x} = \frac{N_x}{2 \pi \sigma_s}
$$

$$
\sigma_{f_y} = \frac{N_y}{2 \pi \sigma_s}
$$

gde su $N_x$ i $N_y$ dimenzije slike u x i y pravcu, dok $\sigma_s$ predstavlja standardnu devijaciju filtra u prostornom domenu.


````python
def calc_freq_sigma(sigma_s, img):
    Ny , Nx = img.shape
    #izotropan filtar sigma_s = sigma_s_x = sigma_s_y
    sigma_x_f, sigma_y_f = Nx / (2 * np.pi * sigma_s), Ny / (2 * np.pi * sigma_s)
    return sigma_x_f, sigma_y_f
````

````python
def filter_c(img_org, sigma_x, sigma_y):
    """funkcija koja filtrira sliku
    1. prebaci sliku u fft domen 
    2. pomnozi sliku sa filtrom u fft domenu 
    3. vrati u prostorni domen sliku 
    4. iseci na (0, 1) opseg ako je neka vrednost izvan opsega
    """
    [Ny, Nx] = np.shape(img_org)

    img = img_org.copy()
    # slika u frekv domenu
    img_fft = np.fft.fftshift(np.fft.fft2(img))

    # filtar
    filt_freq = lpfilter_anisotrop("gaussian", Ny, Nx, sigma_x, sigma_y)

    # filtrirana slika

    img_fft_filt = img_fft * filt_freq

    # vrati nazad u realni domen
    img_filt = np.real(np.fft.ifft2(np.fft.ifftshift(img_fft_filt)))
    img_filt = np.clip(img_filt, 0, 1)

    return img_filt
````

Funkcija `filter_and_plt_img` primenjuje frekvencijsko filtriranje slike za **više vrednosti parametra** `sigma_s`.  
Za svaku vrednost `sigma_s` izračunavaju se odgovarajući parametri filtra u frekvencijskom domenu (`sigma_x_f`, `sigma_y_f`).  
Zatim se slika filtrira pomoću funkcije `filter_c`, a rezultati se prikazuju korišćenjem funkcije `plot_img`.  
Na ovaj način omogućeno je poređenje uticaja različitih vrednosti `sigma_s` na rezultat filtriranja.


````python
def filter_and_plt_img(img, sigma_s):
    sigma_x_f_256, sigma_y_f_256 = calc_freq_sigma(sigma_s, img)
    for sigma_x, sigma_y in zip(sigma_x_f_256, sigma_y_f_256):
        filtered_img = filter_c(img, sigma_x, sigma_y)
        plot_img(img, filtered_img)
````

Prve dve funkcije, `mean_value_spatial` i `mean_value_freq`, služe za merenje **prosečnog vremena izvršavanja filtriranja** u prostornom, odnosno frekvencijskom domenu.  
Za zadati broj ponavljanja (`num_of_times`), svaka funkcija više puta izvršava odgovarajući filter, meri vreme trajanja svakog izvršavanja i na kraju vraća srednju vrednost vremena.

Funkcija `spatial_filtering` koristi `mean_value_spatial` kako bi za niz vrednosti `sigma` izračunala prosečna vremena filtriranja u prostornom domenu.  
Za svaku vrednost parametra `sigma` rezultat se smešta u niz `times`, koji predstavlja zavisnost vremena izvršavanja od širine filtra.

Analogno tome, funkcija `freq_filtering` računa prosečna vremena filtriranja u frekvencijskom domenu.  
Za svaku vrednost `sigma` najpre se izračunavaju odgovarajući parametri filtra u frekvencijskom domenu (`sigma_x` i `sigma_y`), nakon čega se poziva funkcija `mean_value_freq`.  
Na ovaj način omogućeno je direktno poređenje brzine filtriranja u prostornom i frekvencijskom domenu za različite vrednosti parametra `sigma`.


````python
def mean_value_spatial(img, sigma, num_of_times): 
    mean_values = []
    for i in range(0, num_of_times): 
        start = time.time()
        ski.filters.gaussian(img, sigma = sigma)
        end = time.time()
        interval = end - start 
        mean_values.append(interval)
    return np.mean(mean_values)
````

````python
def mean_value_freq(img, sigma, num_of_times, sigma_x, sigma_y):
    mean_values = []
    for i in range(0, num_of_times): 
        start = time.time()
        filter_c(img, sigma_x, sigma_y)
        end = time.time()
        interval = end - start 
        mean_values.append(interval)
    return np.mean(mean_values)
````

````python
def spatial_filtering(img, sigmas, num_of_times):
    times = np.zeros(len(sigmas))
    for i, sigma in enumerate(sigmas):
        times[i] = mean_value_spatial(img, sigma, num_of_times)
    return times    
````

````python
def freq_filtering(img, sigmas, num_of_times):
    times = np.zeros(len(sigmas)) 
    Ny, Nx = img.shape
    for i, sigma in enumerate(sigmas): 
        sigma_x = Nx / (2 * np.pi * sigma)
        sigma_y = Ny / (2 * np.pi * sigma)
        times[i] = mean_value_freq(img, sigma, num_of_times, sigma_x, sigma_y)
    return times 
````

Pošto je slika data u **sRGB** domenu, prvi korak koji je potrebno obaviti jeste njena konverzija iz sRGB domena u **linearni domen**.

Ako je gamma vrednost definisana kao $$ \gamma = \frac{1}{2.2}$$ tada bi se pri prelasku iz linearnog domena u sRGB domen svaka vrednost stepenovala sa  $$\frac{1}{2.2}.$$ Iz tog razloga, obrnuti proces — prelazak iz sRGB u linearni domen — podrazumeva stepenovanje sa 2.2.

Zbog toga je uvedena funkcija **`srgb_to_linear`**, koja omogućava pravilnu transformaciju vrednosti piksela u linearni prostor.

Na kraju obrade, neophodno je sliku iz linearnog domena ponovo vratiti u **sRGB domen**, što se postiže primenom funkcije **`linear_to_srgb`**.


````python
def srgb_to_linear(img):
    img = np.clip(img, 0, 1)
    linear_img = img ** 2.2
    return linear_img
````

````python
def linear_to_srgb(img):
    img = np.clip(img, 0, 1)
    srgb_img = img ** (1 / 2.2)
    return srgb_img
````

Funkcija ```deconvolute_channel``` vrši dekonvoluciju jednog kanala slike u frekvencijskom domenu. Ulazna slika se najpre transformiše u frekvencijski domen pomoću 2D FFT-a. Kernel zamućenja se zatim proširuje nulama kako bi imao iste dimenzije kao slika i takođe se transformiše u frekvencijski domen.

Dekonvolucija se realizuje deljenjem spektra zamućene slike sa spektrom kernela, uz primenu regularizacionog faktora koji smanjuje uticaj šuma i stabilizuje proces. Nakon toga se rezultat vraća u prostorni domen pomoću inverznog FFT-a, uzimajući realni deo signala.

Na kraju, vrednosti piksela se ograničavaju na opseg \([0, 1]\) i izlazna slika se vraća na originalne dimenzije.

````python
def deconvolute_channel(img, kernel):
    orig_height, orig_width = img.shape
    img_fft = np.fft.fftshift(np.fft.fft2(img))
    k = 1e-5

    kernel_height, kernel_width = kernel.shape
    H, W = img.shape[:2]
    kernel_padded = np.zeros((H, W))
    kernel_padded[:kernel_height, :kernel_width] = kernel

    kernel_fft = np.fft.fftshift(np.fft.fft2(kernel_padded))

    W = (abs(kernel_fft) ** 2) / (abs(kernel_fft) ** 2 + k)

    # zamucena slika / faktor zamucenja
    img_fft_est = (img_fft / kernel_fft) * W

    img_est = np.real(np.fft.ifft2(np.fft.ifftshift(img_fft_est)))
    img_est = np.clip(img_est, 0, 1)

    return img_est[:orig_height, :orig_width]
````

funkcija ```deconvolute``` samo crta dve slike i prikazuje ih.

````python
def deconvolute(img, kernel):
    if img.ndim != 3: return;
    output_img = np.zeros_like(img)
    for channel in range(img.shape[2]):
        output_img[:,:, channel] = deconvolute_channel(img[:,:, channel], kernel)
    return output_img
````

````python
def plt_img(img1, img2, title2):
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12,8), dpi=90)
    ax = axes.ravel()

    ax[0].imshow(img1); ax[0].set_title('Ulazna slika')
    ax[1].imshow(abs(img2), cmap='gray'); ax[1].set_title(title2);
    plt.show()
````

````python
def vertical_notch_mask_mesh(Ny, Nx, notch_width=3, dc_radius=8):
    """
    Vertikalni notch filter koristeći meshgrid koordinatni sistem.

    Ny, Nx      : dimenzije FFT-a (height, width)
    notch_width  : širina vertikalne linije
    dc_radius    : koliko piksela oko centra ne diramo
    """
    # horizontalne i vertikalne koordinate centrirane
    if Ny % 2 == 0:
        y = np.arange(0, Ny) - Ny/2 + 0.5
    else:
        y = np.arange(0, Ny) - (Ny-1)/2

    if Nx % 2 == 0:
        x = np.arange(0, Nx) - Nx/2 + 0.5
    else:
        x = np.arange(0, Nx) - (Nx-1)/2

    X, Y = np.meshgrid(x, y)
    D = np.sqrt(X**2 + Y**2)

    # vertikalni notch (Gaussian po X)
    mask = 1 - np.exp(-(X**2) / (2 * notch_width**2))

    # sačuvaj DC oko centra
    mask[D < dc_radius] = 1.0

    return mask
````

````python
def apply_vertical_notch_filter(img, notch_width = 3, dc_radius=10):
    """
    Primeni vertikalni notch filter na linearni (FFT) kanal slike.

    image_linear : slika u linearnom domenu (H,W,3)
    
    return: vrati mi sliku filtriranu notch filtrom
    """
    filtered_image = np.zeros_like(img)
    for channel_index in range(img.shape[2]):
        # FFT kanala
        channel_fft = np.fft.fftshift(np.fft.fft2(img[:,:,channel_index]))
        
        Ny, Nx = channel_fft.shape
        # napravim masku
        notch_mask = vertical_notch_mask_mesh(Ny, Nx, notch_width, dc_radius)

        # filtriram
        filtered_fft = channel_fft * notch_mask

        # inversna FFT
        filtered_image[:,:,channel_index] = np.real(np.fft.ifft2(np.fft.ifftshift(filtered_fft)))

    return np.clip(filtered_image, 0, 1)
````

Dekonvolucija se sprovodi u frekvencijskom domenu, korišćenjem poznatog kernela zamućenja, uz odgovarajuću regularizaciju radi smanjenja uticaja šuma, Filtar koji koristimo je Wienerov filtar koji će lepo isfilttrirati šum na slici.  

Dobijena slika se zatim dodatno obrađuje primenom **notch filtra**, koji se koristi za potiskivanje neželjenih frekvencijskih komponenti, najčešće periodičnog šuma ili interferencija prisutnih u slici.Mogao je biti i isfiltriran i deo koji ima horizontalnu liniju u spektru ali ako bismo njega isfiltrirali i bitne komponente sa slike bi nam bile upropaštene.

Nakon završetka svih operacija u linearnom domenu, slika se ponovo konvertuje iz linearnog u **sRGB (RGB) domen**, kako bi bila pogodna za prikaz na standardnim uređajima. Time se završava kompletan proces obrade slike.

Kao što je opisano u poglavlju, prvo što se treba uraditi jeste napisati funkciju `estimate_noise_var`.
Implementirane su tri funkcije: `estimate_noise_var` kao glavna funkcija, `calc_var` kao pomoćna funkcija za računanje varijanse i `estimate_win` za određivanje odgovarajuće veličine prozora.

Funkcija `calc_var` služi za računanje varijanse signala. Lokalna varijansa se može izračunati pomoću sledeće formule:

$$
\mathrm{Var}(X) = \mathbb{E}(X^2) - \left(\mathbb{E}(X)\right)^2
$$

gde je $\mathbb{E}(X)$ srednja vrednost signala, dok $\mathbb{E}(X^2)$ predstavlja srednju vrednost kvadrata signala unutar posmatranog prozora. Ovaj izraz predstavlja prvi deo implementacije funkcije.

Drugi deo funkcije predstavlja procenu varijanse pomoću histograma. Najpre se izračunava histogram vrednosti signala, zatim se određuju centri histogram-binova. Kao reprezentativna vrednost uzima se maksimalna vrednost histograma, odnosno vrednost u centru histogram-bina sa najvećom učestalošću. Na osnovu ovih koraka dobija se procena varijanse šuma, koja se na kraju vraća kao izlaz funkcije.

````python
def calc_var(noise_img, win):
    '''
    :param noise_img: slika sa sumom
    :param win: velicina prozora
    :return: varijansa = srednja vrednost bina sa maksimalnom vrednoscu onog bina koji ima histogram
    binovi_centri = [1.5, 2.5, 3.5, 4.5], histogram_vrednosti = [1,2,3,4]  varijansa = 4
    '''
    local_mean = ndi.uniform_filter(noise_img, size=win)
    local_sqr_mean = ndi.uniform_filter(noise_img ** 2, size=win)

    local_var = local_sqr_mean - local_mean ** 2

    # vrednost varijanse prethodne
    hist, bin_edges = np.histogram(local_var, bins=256)

    # indeks maks bina prethodne
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

    var = bin_centers[np.argmax(hist)]

    return var
````

Funkcija `estimate_win` računa varijansu za prethodnu veličinu prozora i varijansu za trenutnu veličinu prozora.
Računanje varijansi, a samim tim i određivanje veličine prozora koji zadovoljava uslove, prestaje kada je ispunjena sledeća nejednakost:

$$
\mathrm{Var}(\text{currWin}) - \mathrm{Var}(\text{prevWin}) < \text{tol}
$$

Ukoliko ova nejednakost ne može biti ispunjena, a dostigne se maksimalna dozvoljena veličina prozora, kao konačno rešenje se uzima maksimalna veličina prozora, a odgovarajuća varijansa se prihvata kao rezultat.


````python

def estimate_win(noise_img, min_win, max_win,curr_win, prev_win,
                 tol = 0.001):
    #varijansa prethodne
    var_prev = calc_var(noise_img, prev_win)

    while curr_win <= max_win:

        var_curr = calc_var(noise_img, curr_win)

        if abs(var_curr - var_prev) < tol:
            return curr_win

        var_prev = var_curr
        curr_win += 2

    return max_win
````

Funkcija `estimate_noise_var` proverava vrednost parametra `win`. Ukoliko je veličina prozora jednaka nuli, funkcija poziva `estimate_win`, koja automatski određuje odgovarajuću veličinu prozora na osnovu promene varijanse.

U slučaju da je zadati parametar `win` različit od nule, ta vrednost se direktno koristi kao veličina prozora za dalje računanje.

Nakon određivanja veličine prozora, funkcija računa varijansu šuma za dati prozor korišćenjem prethodno definisane funkcije za računanje varijanse.

Originalno, funkcija `estimate_noise_var` je bila predviđena da vraća samo procenjenu varijansu šuma. Međutim, zbog zahteva iz tačaka b) i c), funkcija je izmenjena tako da kao izlaz vraća i procenjenu varijansu i odgovarajuću veličinu prozora.

````python
def estimate_noise_var(noise_img, win = 0, tol = 0.001):
    min_win, max_win = 3, 51

    if win == 0:
        prev_win, curr_win = min_win, min_win + 2
        win = estimate_win(noise_img, min_win, max_win, curr_win, prev_win)

    var = calc_var(noise_img, win)
    return var, win
````

````python
data_dir = "../sekvence/"
lena_img = ski.util.img_as_float(ski.io.imread(data_dir + "lena.tif"))
````

````python
def local_hist(noise_img, curr_win):
    '''
    vraca maks vrednost histograma varijansi
    '''
    kernel = np.ones((curr_win, curr_win)) / (curr_win ** 2)

    local_mean = ndi.uniform_filter(noise_img, size=curr_win)
    local_sqr_mean = ndi.uniform_filter(noise_img ** 2, size=curr_win)

    local_var = local_sqr_mean - local_mean ** 2

    # histogram varijanse
    hist, bin_edges = np.histogram(local_var, bins = 64)

    hist_max = np.max(hist)

    return hist_max
````

Funkcija `get_result` vrši pet(jer se trazilo od 5 do 10 merenja) uzastopnih merenja varijanse i maksimalnih vrednosti histograma za svaku veličinu prozora. U svakoj od pet iteracija formira se lista koja sadrži varijanse za sve moguće veličine prozora, počev od minimalne do maksimalne vrednosti prozora.

Nakon što se izvrši svih pet merenja, za svaku veličinu prozora posebno se analiziraju dobijeni rezultati. Naime, za dati prozor (npr. prozor veličine 3), uzimaju se sve izmerene varijanse iz pet iteracija koje pripadaju datom prozoru sa veličinom 3 i računa se njihova srednja vrednost, čime se dobija srednja varijansa za tu veličinu prozora. Nad istim skupom vrednosti zatim se računa standardna devijacija.Na analogan način obrađuju se i maksimalne vrednosti histograma za ovu veličinu prozora.  

Ovaj postupak se ponavlja za svaku moguću veličinu prozora.

Na kraju, funkcija `get_result` vraća listu srednjih vrednosti varijanse, listu standardnih devijacija, kao i krajnji histogram za sve posmatrane veličine prozora.

````python
def get_result(min_window, max_window):
    ctr, num_of_measurements = 0, 5
    
    # broj prozora
    num_of_cols = ((max_window - min_window) // 2) + 1
    
    # privremene vrednosti varijansi
    temp_vals = np.zeros((num_of_measurements, num_of_cols))
    
    # privremene vrednosti histograma (5x listi po jedna vrednost za svaku velicinu prozora u listi)
    temp_hist_vals = np.zeros((num_of_measurements, num_of_cols))

    # [[varijanse prve], [varijanse druge], [varijanse trece] , [varijanse cetvrte], [varijanse pete]]
    for measure in range(num_of_measurements):
        # za svako merenje:

        # 1. dodaj sum na sliku lene
        lena_gauss_noise_0p01 = ski.util.random_noise(lena_img, mode="gaussian", mean=0, var=0.01)

        # 2. [proceni varijansu za svaki prozor]
        temp_vals[ctr] = [estimate_noise_var(lena_gauss_noise_0p01, window)[0] for window in range(min_window, max_window + 1, 2)]

        # 3. [vrednosti histograma za svaki prozor]
        temp_hist_vals[ctr] = [local_hist(lena_gauss_noise_0p01, window) for window in range(min_window, max_window + 1, 2)]
        ctr += 1

    final_var = np.zeros(num_of_cols)
    final_stand_dev = np.zeros(num_of_cols)
    final_hist = np.zeros(num_of_cols)

    # izracunam standardne devijacije i varijanse
    for i in range(0, num_of_cols):
        final_var[i] = np.mean(temp_vals[:, i])
        final_stand_dev[i] = np.std(temp_vals[:, i])
        final_hist[i] = np.mean(temp_hist_vals[:, i])
    return final_var, final_stand_dev, final_hist
````

Nakon dobijanja rezultata pozivom funkcije `get_result`, dobijene su tri vrednosti: srednje varijanse šuma (`vars`), standardne devijacije (`standard_devs`) i usrednjene maksimalne vrednosti histograma (`hist_vals`). Veličine prozora za koje su rezultati računati definisane su nizom `windows`, koji obuhvata neparne vrednosti od minimalne veličine prozora 3 do maksimalne veličine prozora 51.

Na prvom grafiku prikazana je zavisnost procenjene varijanse šuma od veličine prozora. Tačke na grafiku predstavljaju srednje vrednosti varijanse, dok su vertikalne greške prikazane pomoću standardne devijacije. Ovakav prikaz omogućava uvid u stabilnost procene varijanse u zavisnosti od veličine prozora, kao i u rasipanje rezultata dobijenih ponovljenim merenjima.

Na drugom grafiku prikazana je zavisnost usrednjenih maksimalnih vrednosti histograma od veličine prozora. Ovaj grafikon ilustruje kako se dominantna vrednost histograma menja sa promenom veličine prozora, što daje dodatnu informaciju o ponašanju šuma i distribuciji intenziteta u signalu.


````python
vars, standard_devs, hist_vals = get_result(3, 51)
windows = np.arange(3, 52, 2)  # min_window=3, max_window=51, korak=2


# PRVI GRAFIK
plt.figure(figsize=(8,5))
plt.errorbar(windows, vars, yerr=standard_devs, fmt='-o', capsize=5, ecolor='blue', color='red', label = "Procena šuma")
plt.xlabel("Veličina prozora")
plt.ylabel("Varijansa šuma")
plt.title("Varijansa ± standardna devijacija")
plt.legend()
plt.grid(True)
plt.show()

# DRUGI GRAFIK
plt.figure(figsize=(8,5))
plt.plot(windows, hist_vals, '-o', color='green')
plt.xlabel("Veličina prozora")
plt.ylabel("Usrednjene  maks vrednosti histograma")
plt.title(r"$hist = f[window\_size]$")
plt.grid(True)
plt.show()
````

Funkcija `test_automatic` za svaku zadatu vrednost varijanse izvršava pet nezavisnih merenja. Nakon sprovedenih pet merenja, na osnovu dobijenih rezultata računa se procenjena varijansa šuma, kao i odgovarajuća prosečna veličina prozora.

Greška procene definiše se kao razlika između izračunate i stvarne varijanse, odnosno:

$$
\text{greška} = \text{Var}_{\text{izračunata}} - \text{Var}_{\text{prava}}
$$

Na osnovu dobijenih varijansi računa se standardna devijacija.  
Srednja vrednost svih veličina prozora koje su korišćene za datu varijansu se takodje racuna za merenja.

Parametar `tol` direktno utiče na izbor veličine prozora. Manje vrednosti parametra `tol` zahtevaju veće veličine prozora kako bi uslov konvergencije bio zadovoljen, dok veće vrednosti parametra `tol` dovode do izbora manjih veličina prozora.

Na kraju, funkcija `test_automatic` kao izlaz vraća srednju vrednost procenjene varijanse, standardnu devijaciju, greške i srednju veličinu prozora.


````python
def test_automatic(tol):
    # var : 0.001 -> 0.1
    # za var = 0.001: 5x izmerim varijansu pomocu automatic window, zatim izracunam srednju vrednost i std za var = 0.001 za odredjeni prozor
    # var = 0.1 5x izmerim varijansu pomocu automatic window, mean 5x vrednosti , std 5x vrednosti
    num_of_measurements = 5
    windows = np.zeros(num_of_measurements)
    all_windows = {}
    random_values = np.linspace(0.001, 0.1, 10)
    var, std_deviations, errors, temp_variances = {}, {}, {}, np.zeros(num_of_measurements)
    for val in random_values:
        # za svaku varijansu:
        for i in range(num_of_measurements):
            # za svako merenje:
            # 1. stavi slucajan sum na varijansu
            # 2. izracunaj varijansu, prozor koji je odredjen
            lena_rnd_img = ski.util.random_noise(lena_img, mode="gaussian", mean=0, var=val)
            temp_variances[i], windows[i] = estimate_noise_var(lena_rnd_img,  0, tol = tol)

        # 3. var[0.001] : srednja_vrednost_5xvarijansi
        # 4. greska[0.001] = dobijena - prava
        # 5. standardne devijacije
        # 6. izracunam prosecnu velicinu prozora iz 5x merenja za jednu varijansu

        var[val] = np.mean(temp_variances)
        errors[val] = abs(var[val] - val)
        std_deviations[val] = np.std(temp_variances)
        all_windows[val] = np.mean(windows)
    return var, std_deviations, errors, all_windows
````

## a)

Funkcija `box_filter` primenjuje **kockasti filter** na ulaznu sliku `img`.  
Veličina filtra određena je parametrom `R`, koji definiše dimenziju prozora u kom se racuna srednja vrednost.

Filter se računa pomoću funkcije `uniform_filter`, pri čemu se koristi `mode="reflect"` kako bi se ivice slike tretirale refleksijom, kao da ima ogledala pa sa druge strane su brojevi.

Rezultat funkcije je filtrirana slika gde je svaki piksel zamenjen prosečnom vrednošću unutar prozora veličine `R`.


````python
def box_filter(img, R):
    return uniform_filter(img, size = R, mode = "reflect")
````

Ako se pogleda sema rada **guided filtara** u radu *Fast self-guided filter with decimated box filters*, može se primetiti da oba filtra imaju jedan zajednički deo, a to je **računanje koeficijenata**.  

Sam postupak računanja koeficijenata prikazan je u funkciji `compute_coefficients`, dok je odgovarajuća slika prikazana nakon pokretana celije ispod.  
Ova slika ilustruje kako se koeficijenti koriste u procesu filtriranja i daje vizuelni uvid u zajedničku strukturu oba filtra.


````python
img = ski.io.imread("../sekvence/ss.png")
ski.io.imshow(img)
````

Funkcija `compute_coefficients` računa koeficijente \(a\) i \(b\) za **guided filtre**. Postupak se sastoji od nekoliko koraka:

1. Kvadrira se ulazna slika `img`.
2. Računa se lokalna srednja vrednost kvadrirane slike koristeći `box_filter` (leva grana formule).
3. Računa se lokalna srednja vrednost originalne slike `img` (desna grana formule).
4. Kvadrat srednje vrednosti originalne slike se računa.
5. Varijansa unutar prozora se dobija oduzimanjem kvadrata srednje vrednosti od srednje vrednosti kvadrata:  
   $$
   \text{varWin} = E(I^2) - (E(I))^2
   $$
6. Koeficijent \(a\) se računa kao:  
   $$
   a = \frac{\text{varWin}}{\text{varWin} + \epsilon}
   $$
7. Koeficijent \(b\) se računa kao:  
   $$
   b = E(I) - a \cdot E(I)
   $$
8. Na kraju se za oba koeficijenta računa lokalna srednja vrednost pomoću `box_filter`, i funkcija vraća `a_mean` i `b_mean`.

Ovi koeficijenti se dalje koriste u guided filteru za linearnu kombinaciju originalne slike i filtrirane komponente.


````python
def compute_coefficients(img, mask, epsilon):
    # 1 kvadriram sliku
    img_squared = img ** 2

    # 2 usrednjim kvadratnu sliku (leva grana)
    img_sqr_mean_left = box_filter(img_squared, mask)

    # 3 usrednjim obicnu sliku ( desna grana )
    img_mean = box_filter(img, mask)

    # 4 dignem obicnu sliku usrednjenu na kvadrat
    img_sqr_mean_right = img_mean ** 2

    # 5 saberi levu stranu oduzmi desnu stranu
    var_win = img_sqr_mean_left - img_sqr_mean_right

    # 6 izracunam koeficijent a
    a = var_win / (var_win + epsilon)

    # 7 izracunam b to je ovo desno dole sto se sabira
    b = img_mean - a * img_mean

    # 8 izracunam srednju vrednost od a
    a_mean, b_mean = box_filter(a, mask), box_filter(b, mask)

    return a_mean, b_mean
````

Funkcija `self_guided_dos_filter` primenjuje **self-guided filter** na ulaznu sliku `img`.  
Ako je slika u boji, filtriranje se vrši po kanalu, to jest funkcija se rekurzivno poziva,a rezultati se kombinuju u izlaznu sliku.  

Prvo se računaju koeficijenti `a_mean` i `b_mean` pomoću funkcije `compute_coefficients`.  
Zatim se množi koeficijent `a_mean` sa originalnom slikom i sabira sa `b_mean` da bi se dobila filtrirana izlazna slika.  

Postupak je kao na slici koja se dobija nakon pokretanja ćelije ispod.



````python
img = ski.io.imread("../sekvence/dos_guided_filter.png")
ski.io.imshow(img)
````

Kao što se vidi, koeficijenti **$a$** i **$b$** za **box filter** su već izračunati. Jedino što preostaje jeste da se prati prikazana šema korak po korak i na osnovu nje napiše odgovarajuća funkcija koja implementira filter. Na taj način se osigurava da implementacija odgovara teorijskom modelu i prethodno definisanim parametrima.


````python
def self_guided_dos_filter(img, R, epsilon):
    if img.ndim == 3:
        output = np.zeros_like(img)
        for c in range(img.shape[2]):
            output[:, :, c] = self_guided_dos_filter(img[:, :, c], R, epsilon)
        return output

    a_mean, b_mean = compute_coefficients(img, R, epsilon)

    # 9 a_mean x img
    img_mean_product = a_mean * img

    # 10 izlazna slika
    output_img = b_mean + img_mean_product

    return output_img
````

## b)

Funkcija `fast_self_guided_dos_filter` implementira **bržu verziju self-guided filtera** koristeći decimaciju slike.  
Ako je ulazna slika u boji, filtriranje se radi po kanalu, a rezultati se kombinuju u izlaznu sliku.  

Prvo se računa smanjen prozor `K` na osnovu originalne veličine prozora `R` i decimacije `S`.  
Zatim se ulazna slika smanjuje (`decimated_img`) kako bi se ubrzalo računanje koeficijenata.  

Koeficijenti `a_mean` i `b_mean` se računaju na smanjenoj slici pomoću `compute_coefficients`.  
Nakon toga se koeficijenti interpoliraju nazad na originalne dimenzije slike (`zoom`) i seku na tačne dimenzije.  

Na kraju se filtrirana izlazna slika dobija linearnom kombinacijom `a_mean_up * img + b_mean_up`.  
Ova metoda značajno ubrzava filtriranje, a i dalje zadržava detalje i ivice slike zahvaljujući guided principu.

Cela ova metoda je napisana prateći šemu koja se dobije na slici nakon pokretanja ćelije ispod korak po korak.


````python
img = ski.io.imread("../sekvence/dos_fast_guided_filter.png")
ski.io.imshow(img)
````

````python
def fast_self_guided_dos_filter(img, R, epsilon, S):
    if img.ndim == 3:
        output = np.zeros_like(img)
        for c in range(img.shape[2]):
            output[:, :, c] = fast_self_guided_dos_filter(img[:, :, c], R, epsilon, S)
        return output

    # 1 izracunam K
    K = R / S

    # 2 smanjim dimenziju ulazne slike
    decimated_img = img[::S, ::S, ...]

    # 3 izracunam koeficijente iz funkcije 
    a_mean, b_mean = compute_coefficients(decimated_img, K, epsilon)

    # 4 sad vratim nazad sliku na originalne dimenzije
    a_mean_up = zoom(a_mean, zoom=S, order=1)
    b_mean_up = zoom(b_mean, zoom=S, order=1)

    # 5 isecem/popunim do originalnih dimenzija (inace dobijem python exception)
    h, w = img.shape[:2]
    a_mean_up = a_mean_up[:h, :w]
    b_mean_up = b_mean_up[:h, :w]

    #6 izracunam izlaznu sliku
    output_img = a_mean_up * img + b_mean_up

    return output_img
````

Funkcija `calc_psnr` računa **PSNR (Peak Signal-to-Noise Ratio)** između slika dobijenih korišćenjem `dos_guided` i `fast_dos_guided` algoritama.  

Ako je ulazna slika u boji (RGB), PSNR se računa posebno po svakom kanalu, a zatim se te tri vrednosti prosečno kombinuju da bi se dobila ukupna vrednost.  
U slučaju da je slika grayscale, PSNR se računa klasičnim pozivom funkcije, direktno na celoj slici.

Ovaj postupak omogućava poređenje kvaliteta filtriranih slika u odnosu na originalnu sliku.


````python
def calc_PSNR(img, img_dos_guided, img_fast):
    # PSNR između slika
    if img.ndim == 3:
        # Za RGB, računaj po kanalima
        psnr_val = np.mean([
            psnr(img_dos_guided[:, :, c], img_fast[:, :, c], data_range=1.0)
            for c in range(3)
        ])
    else:
        psnr_val = psnr(img_dos_guided, img_fast, data_range=1.0)

    return psnr_val
````

Funkcija `test_one_image` vrši testiranje **guided** i **fast self-guided DOS filtra** na jednoj slici `img`.  

Prvo se meri vreme filtriranja običnim `self_guided_dos_filter` i čuva se kao `t_img_guided`.  
Zatim se meri vreme filtriranja brzim `fast_self_guided_dos_filter` algoritmom i čuva se kao `t_img_fast`.

Funkcija vraća vreme izvršavanja oba filtra i vrednost PSNR, što omogućava poređenje kvaliteta i brzine oba algoritma.


````python
def test_one_image(img, R_val, eps, s = 2):
    start = time.time()
    img_dos_guided = self_guided_dos_filter(img, R_val, eps)
    end = time.time()
    t_img_guided = end - start

    # Brzi filter
    start = time.time()
    img_fast = fast_self_guided_dos_filter(img, R_val, eps, 2)
    end = time.time()
    t_img_fast= end - start

    PSNR_VAL = calc_PSNR(img, img_dos_guided, img_fast)

    return t_img_guided, t_img_fast, PSNR_VAL
````

Funkcija `create_pyramid` kreira **piramidu slika** gde je svaka naredna slika
dobijena **smanjenjem rezolucije za faktor 2** u odnosu na prethodnu.

Proces se odvija tako što se prvo odabira svaka druga vrednost po visini i širini, čime se rezolucija prepolovi.

Sve generisane slike se čuvaju u listi, a na kraju se redosled liste
invertuje tako da:
- prva slika u listi ima **najmanju rezoluciju** (najgrublji nivo),
- poslednja slika ima **originalnu rezoluciju**.

Ovakav raspored omogućava obradu slike po principu *coarse-to-fine*,
što je posebno pogodno za procenu kernela i iterativno uklanjanje zamućenja.


````python
def create_pyramid(img, num_scales):
    pyr = [img]
    for _ in range(1, num_scales):
        img = ski.filters.gaussian(img, sigma=1)
        img = img[::2, ::2]
        pyr.append(img)
    return pyr[::-1]  # coarse to fine
````

````python
def init_LK(pyramid, ksize):
    L = pyramid[0].copy()
    K = np.zeros((ksize, ksize))
    K[ksize//2, ksize//2] = 1.0
    return L, K
````

### 1. Bilateralni filter (`bilateral`)

Ova funkcija primjenjuje bilateralni filter na ulaznu sliku. Cilj bilateralnog filtriranja je uklanjanje šuma uz istovremeno očuvanje ivica. Za razliku od klasičnih linearnih filtara, bilateralni filtar uzima u obzir i prostornu udaljenost piksela i razliku u intenzitetu, zbog čega se ivice ne zamućuju. Koristi se kao metoda za edge-preserving denoising, naročito kada je važno zadržati strukturalne detalje slike.

---

### 2. Shock filter (`shock`)

Ova funkcija implementira nelinearni shock filter koji služi za izoštravanje ivica na slici. Zasniva se na kombinaciji Laplasovog operatora i magnitude gradijenta. Laplasijan određuje da li se piksel nalazi u svijetloj ili tamnoj regiji u odnosu na okolinu, dok gradijent daje informaciju o jačini ivice. Rezultat je pojačavanje diskontinuiteta i stvaranje oštrijih, jasnije definisanih ivica, uz kontrolu jačine efekta pomoću vremenskog koraka.

---

### 3. Pragovanje gradijenta (`gradient_threshold`)

Ova funkcija izdvaja najznačajnije ivice na slici tako što računa gradijent i zadržava samo one delove sa najvećom magnitudom. Nakon izračunavanja jačine gradijenta, određuje se prag na osnovu zadatog procenta koji definiše koliko najjačih ivica treba sačuvati. Slabiji gradijenti se uklanjaju, čime se dobija redak prikaz ivica. Ovaj pristup je koristan u detekciji ivica, redukciji šuma i u variacionim metodama obrade slike.


````python
def bilateral(L):
    return ski.restoration.denoise_bilateral(L, sigma_spatial=2, sigma_color=0.05)
````

````python
def shock(L, dt):
    lap = ski.filters.laplace(L)
    grad = np.sqrt(np.square(ski.filters.sobel_h(L)) + np.square(ski.filters.sobel_v(L)))
    return L - dt * np.sign(lap) * grad
````

````python
def gradient_threshold(L, thresh):
    """Threshold gradient adaptively, decaying thresh per iteration."""
    Px = ski.filters.sobel_h(L)
    Py = ski.filters.sobel_v(L)
    mag = np.sqrt(Px**2 + Py**2)
    thresh_val = np.percentile(mag, (1 - thresh) * 100)  # keep top fraction
    mask = mag > thresh_val
    return Px * mask, Py * mask
````

````python
def normalize_kernel(K):
    K[K < K/20] = 0
    s = K.sum()
    if s > 0:
        K /= s
    return K
````

### Procena kernela zamućenja (`estimate_kernel`)

Ova funkcija služi za procenu kernela zamućenja (blur kernel / PSF) na osnovu gradijenata oštre slike i zamućene slike. .

Na početku se iz zamućene slike izdvajaju samo najjači gradijenti kako bi se smanjio uticaj šuma i slabih ivica. Zatim se procena kernela vrši u **Fourierovom domenu**, gde se problem formulše kao regularizovani least-squares odnos između gradijenata oštre i zamućene slike. Parametar regularizacije sprečava numeričku nestabilnost i preuveličavanje šuma.

Dobijeni kernel se vraća u prostorni domen, centrira i zatim se iseca na željenu veličinu. Nakon toga se primenjuje dodatna stabilizacija u vidu Gaussovog prigušenja ka ivicama kernela, čime je cilj da se izvuku kompaktni i glatki kernel oblici. Na kraju se kernel normalizuje tako da ukupna suma ostane kontrolisana.

Rezultat funkcije je stabilan i fizički smislen kernel zamućenja koji najbolje objašnjava odnos između ulaznih gradijenata slike i zamućenog posmatranja.

````python
def estimate_kernel(Px, Py, B, ksize, beta):
    Bx, By = gradient_threshold(B, 0.1)

    nominator = 0
    denominator = beta

    for P, Bgrad in [(Px, Bx), (Py, By)]:
        Pf = np.fft.fft2(P)
        Bf = np.fft.fft2(Bgrad)
        nominator += np.conj(Pf) * Bf
        denominator += np.conj(Pf) * Pf

    K = np.real(np.fft.ifft2(nominator / (denominator + 1e-8)))
    K = np.fft.fftshift(K)

    # Crop to kernel size
    c = np.array(K.shape) // 2
    r = ksize // 2
    K = K[c[0]-r:c[0]+r+1, c[1]-r:c[1]+r+1]

    # Stabilizacija
    sigma = ksize / 4
    yy, xx = np.indices(K.shape)
    cy, cx = K.shape[0] // 2, K.shape[1] // 2
    K *= np.exp(-((yy - cy) ** 2 + (xx - cx) ** 2) / (2 * sigma ** 2))
    K = normalize_kernel(K)

    return K
````

### Konverzija PSF u OTF (`psf2otf`)

Ova funkcija pretvara **PSF (Point Spread Function)** iz prostornog domena u **OTF (Optical Transfer Function)** Fourier domenu. OTF predstavlja frekvencijski odziv sistema i koristi se u dekonvoluciji i filtriranju slike u frekvencijskom domenu.

Funkcija najpre smešta PSF u matricu zadate veličine, popunjavajući ostatak nulama (zero-padding), kako bi dimenzije odgovarale slici nad kojom se radi. Zatim se kernel centrira, što je neophodno za ispravnu Fourierovu transformaciju.

Na kraju se primenjuje dvodimenzionalna Fourierova transformacija, čime se dobija OTF. Rezultat se koristi za efikasne konvolucije i dekonvolucije u frekvencijskom domenu.


````python
def psf2otf(psf, shape):
    pad = np.zeros(shape)
    h, w = psf.shape
    pad[:h, :w] = psf
    pad = np.roll(pad, -h//2, 0)
    pad = np.roll(pad, -w//2, 1)
    return np.fft.fft2(pad)
````

### Dekonvolucija slike (`deconvolve`)

Ova funkcija vrši **regularizovanu dekonvoluciju** zamućene slike koristeći poznat kernel zamućenja. Cilj je rekonstrukcija latentne (oštre) slike iz zamućenog posmatranja, uz kontrolu šuma i numeričke nestabilnosti.

Proces se odvija u **Fourierovom domenu**, gde se konvolucija slike i kernela zamućenja može efikasno predstaviti kao množenje u frekvencijskom prostoru. Kernel se najpre transformiše u OTF, dok se zamućena slika prevodi u frekvencijski domen pomoću Fourierove transformacije. Zatim se uvodi skup diskretnih diferencijalnih operatora koji aproksimiraju izvode nultog, prvog i drugog reda, čime se obezbeđuje regularizacija rešenja i smanjenje uticaja šuma.


Svaki diferencijalni operator doprinosi numeričkom i nazivničkom delu rešenja u skladu sa dodeljenom težinom. Time se dobija balans između vernosti podacima i glatkoće rešenja. Parametar regularizacije kontroliše koliko će se potisnuti šum i nestabilne visoke frekvencije.

Na kraju se rezultat vraća u prostorni domen, uzima se realni deo, i vrednosti se ograničavaju na dozvoljeni opseg intenziteta. Rezultat je procena originalne slike.


````python
def deconvolve(B, K, alpha):
    h, w = B.shape

    # 1. Kernel OTF
    Kf = psf2otf(K, (h, w))

    # 2. FFT slike (bez shift-a!)
    Bf = np.fft.fft2(B)

    derivs = {
        "0": np.ones((1, 1)),
        "x": np.array([[-1, 1]]),
        "y": np.array([[-1], [1]]),
        "xx": np.array([[1, -2, 1]]),
        "yy": np.array([[1], [-2], [1]]),
        "xy": np.array([[1, -1], [-1, 1]])
    }

    weights = {"0": 1, "x": 1, "y": 1, "xx": 0.5, "yy": 0.5, "xy": 0.5}

    num = 0
    den = alpha

    for k, v in derivs.items():
        D = np.zeros((h, w))
        v = np.array(v)
        D[:v.shape[0], :v.shape[1]] = v
        D = np.roll(D, -v.shape[0] // 2, 0)
        D = np.roll(D, -v.shape[1] // 2, 1)
        Df = np.fft.fft2(D)
        num += weights[k] * np.conj(Kf * Df) * (Df * Bf)
        den += weights[k] * np.abs(Kf * Df) ** 2

    Lf = num / (den + 1e-8)
    L = np.real(np.fft.ifft2(Lf))
    L = np.clip(L, 0, 1)

    return L
````

Sam postupak algoritma se sastoji od sledećih koraka:

1. **Izgradnja piramide slika**  
   Napravimo piramidu slika, gde se na dnu nalazi slika sa najmanjom rezolucijom, dok je na vrhu slika sa originalnom rezolucijom. Svaka slika u piramidi predstavlja jedan nivo.

2. **Određivanje broja iteracija po nivou**  
   Za svaki nivo piramide definiše se broj iteracija koje će algoritam izvršiti na tom nivou.

3. **Iteracije po nivou**  
   Za svaki nivo piramide vrši se unapred definisan broj iteracija. Svaka iteracija se sastoji iz sledećih koraka:  
   
   3.1 **Prediction korak**  
   U ovom koraku primenjuju se tri metode nad trenutnom slikom:  
   - Bilateralni filter – za uklanjanje šuma uz očuvanje ivica.  
   - Shock filter – za pojačavanje i izoštravanje ivica.  
   - Gradient threshold – za selekciju najjačih gradijenata (ivica).  

   3.2 **Procena kernela zamućenja**  
   Na osnovu trenutne slike i rezultata iz predikcionog koraka, algoritam pokušava da odredi kernel zamućenja (Point Spread Function, PSF) koji najbolje opisuje način na koji je slika zamućena.

   3.3 **Dekonvolucija**  
   Koristeći procenjeni kernel, algoritam pokušava da rekonstruiše originalnu sliku iz zamućene, što predstavlja klasičan problem **blind deconvolution**, s obzirom da je slika data formulom:  
   $$
   B = L * K + N
   $$  
   gde je $B$ zamućena slika, $L$ originalna slika, $K$ kernel zamućenja, a $N$ šum.

4. **Prolazak kroz sve nivoe piramide**  
   Nakon završetka iteracija na jednom nivou, rekonstruisana slika se **upsample-uje** i koristi kao inicijalna slika za sledeći nivo piramide.  
   
   Nakon prolaska kroz sve nivoe piramide, dobija se finalna slika koja je odzamucena korišćenjem ovog algoritma.


````python
def process_one_pyramid_level(L, pyramid_img, num_of_iters, ksize, beta, alpha, dt, sigma_r):
    thresh = 0.1  # initial gradient threshold fraction
    for _ in range(num_of_iters):
        Lb = bilateral(L)
        Ls = shock(Lb, dt)
        Px, Py = gradient_threshold(Ls, thresh)

        K = estimate_kernel(Px, Py, pyramid_img, ksize, beta)
        L = deconvolve(pyramid_img, K, alpha)

        # decay parameters
        dt *= 0.9
        thresh = max(thresh * 0.9, 0.03) # lagano decaying, ne previše

    return L, K
````

````python
def blind_deblur_gray(B, ksize, beta, alpha):
    pyramid = create_pyramid(B, 5)
    L, K = init_LK(pyramid, ksize)

    sigma_r = 0.5
    dt = 1

    for i in range(len(pyramid)):
        L, K = process_one_pyramid_level(L, pyramid[i], 10, ksize, beta, alpha, dt, sigma_r)

        if i < len(pyramid)-1:
            L = ski.transform.resize(L, pyramid[i+1].shape, anti_aliasing=True)
            K[K < 0] = 0
            K /= K.sum() + 1e-8
    return L, K
````

Funkcija **`blind_deblur_gray`** primenjuje opisani algoritam na **grayscale slike**.  

Funkcija **`fast_motion_deblur_rgb`** vrši isti algoritam na **RGB slikama**, ali ne obrađuje svaki kanal posebno, da bi se izbeglo višestruko izvođenje teških proračuna. Umesto toga, RGB slika se prvo konvertuje u **YCbCr** prostor boja, zatim se izračuna **luminansa (Y kanal)** i algoritam se primenjuje samo na nju.  

Nakon što je luminansa odzamucena, rezultat se kombinuje sa originalnim **Cb i Cr kanalima**, i slika se vraća u **RGB domen**. Na taj način se dobija odzamućena RGB slika, a kernel zamućenja se može koristiti za dalju obradu ili vizuelizaciju.


````python
def fast_motion_deblur_rgb(img, ksize, beta, alpha):
    img = ski.util.img_as_float(img)
    ycbcr = ski.color.rgb2ycbcr(img)
    Y = ycbcr[..., 0] / 255.0
    restored_Y, K = blind_deblur_gray(Y, ksize, beta, alpha)
    restored_Y = np.clip(restored_Y, 0, 1)
    ycbcr[..., 0] = restored_Y * 255.0
    restored_rgb = ski.color.ycbcr2rgb(ycbcr)
    restored_rgb = np.clip(restored_rgb, 0, 1)
    return restored_rgb, K
````

### Funkcija `dos_fast_motion_deblurring`

Ova funkcija predstavlja **glavni interfejs** za odzamućivanje slike. Na osnovu dimenzija ulazne slike automatski odlučuje koji algoritam će se primeniti:

- Ako je ulazna slika **RGB** (`ndim == 3`), poziva se funkcija `fast_motion_deblur_rgb`, koja primenjuje algoritam nad **luminansom Y-kanala** u YCbCr prostoru i vraća rekonstruisanu RGB sliku.
- Ako je ulazna slika **grayscale** (`ndim == 2`), koristi se funkcija `blind_deblur_gray`, koja radi isti algoritam direktno nad jednim kanalom.
- Ako slika nije ni RGB ni grayscale, funkcija baca grešku (`TypeError`) i upozorava korisnika da ulaz nije validan.

Funkcija vraća **odzamućenu sliku** istih dimenzija kao ulazna, spremnu za dalju obradu ili vizuelizaciju.


````python
def dos_fast_motion_deblurring(img, ksize=15, beta=5, alpha=0.1):
    if img.ndim == 3:
        output_img, _ = fast_motion_deblur_rgb(img, ksize, beta, alpha)
    elif img.ndim == 2:
        output_img, _ = blind_deblur_gray(img, ksize, beta, alpha)
    else:
        raise TypeError("Error! img must be rgb or grayscale")
    return output_img
````

Na slici ispod prikazana je **originalna slika** sa leve strane, dok se sa desne strane vidi **odzamućena slika** dobijena primenom algoritma. Može se uočiti da algoritam funkcioniše za zadate parametre, iako rezultat nije potpuno idealan, posebno u oblastima sa složenim detaljima ili jačim zamućenjima.


````python
data_dir = "../sekvence/"
img = ski.util.img_as_float(ski.io.imread(data_dir + "blurry_dolls.png"))
img = img[:, :, :3]

result_img = dos_fast_motion_deblurring(img, ksize=11, beta=5, alpha=0.1)



plt.figure(figsize=(14,5))
plt.subplot(1,2,1); plt.title("Blurred"); plt.imshow(img); plt.axis("off")
plt.subplot(1,2,2); plt.title("Restored"); plt.imshow(result_img); plt.axis("off")
plt.show()

````

Analiza grafika u **frekvencijskom domenu** pokazuje da horizontalne linije na slici generišu **vertikalne linije** u spektru, dok vertikalne linije na slici proizvode **horizontalne linije** u spektru.  

Ukoliko bi bilo potrebno ukloniti ovakve smetnje, mogla bi se primeniti **notch filtracija**. Na primer, filtriranjem horizontalnih linija u spektru, vertikalne linije bi nestale sa same slike. Analogno, filtriranjem vertikalnih linija u spektru, uklonile bi se horizontalne linije sa slike. Ovaj princip omogućava selektivno uklanjanje periodicnog šuma ili neželjenih obrazaca u slici.
