### 1.1.1 Prikazivanje signala i frekvencijskog spektra



```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.fft as fft

# Parametri
fs, Ts = 100, 1/100
f1, f2, f3 = 1, 3, 7
t = np.arange(0, 5, Ts)
x = np.cos(2 * np.pi * f1 * t) + 0.5*np.cos(2 * np.pi * f2 * t) + 3 * np.cos(2*np.pi* f3 * t)

# Sampling kontinualnog signala
N1, N2, N3 = 32, 128, 1024
F1, F2, F3 = f1/fs, f2/fs, f3/fs

# Subplot figura
fig, axs = plt.subplots(4, 1, figsize=(8, 10))
plt.subplots_adjust(hspace=1.3, bottom = 0.15)

# 1. Kontinualni signal
axs[0].plot(t, x)
axs[0].set_title("0: Kontinualni signal x(t)")
axs[0].set_xlabel("t [s]")
axs[0].set_ylabel("x(t)")
axs[0].set_ylim([-5, 5])

# 2. FFT za N1
n1 = np.arange(N1)
fk1 = n1 * fs / N1
x1 = np.cos(2 * np.pi * F1 * n1) + 0.5 * np.cos(2 * np.pi * F2 * n1) + 3 * np.cos(2 * np.pi * F3 * n1)
X1 = fft.fft(x1)
axs[1].stem(fk1, np.abs(X1), basefmt='none')
axs[1].set_xlim([0, fs/5])
axs[1].set_title("1: Spektar (N1 = 32)")
axs[1].set_xlabel(r'$f_{k_1}$')
axs[1].set_ylabel(r'$|X[f_{k_3}]|$')

# 3. FFT za N2
n2 = np.arange(N2)
fk2 = n2 * fs / N2
x2 = np.cos(2 * np.pi * F1 * n2) + 0.5 * np.cos(2 * np.pi * F2 * n2) + 3 * np.cos(2 * np.pi * F3 * n2)
X2 = fft.fft(x2)
axs[2].stem(fk2, np.abs(X2), basefmt='none')
axs[2].set_xlim([0, fs/5])
axs[2].set_title("2: Spektar (N2 = 128)")
axs[2].set_xlabel(r'$f_{k_2}$')
axs[2].set_ylabel(r'$|X[f_{k_2}]|$')

# 4. FFT za N3
n3 = np.arange(N3)
fk3 = n3 * fs / N3
x3 = np.cos(2 * np.pi * F1 * n3) + 0.5 * np.cos(2 * np.pi * F2 * n3) + 3 * np.cos(2 * np.pi * F3 * n3)
X3 = fft.fft(x3)
axs[3].stem(fk3, np.abs(X3), basefmt='none')
axs[3].set_xlim([0, fs/5])
axs[3].set_title("3: Spektar (N3 = 1024)")
axs[3].set_xlabel(r'$f_{k_3}$')
axs[3].set_ylabel(r'$|X[f_{k_3}]|$')

plt.show()
```

```python
def computing_frequencies(frequencies,maxPeakIndex,fs,N):
    for i in range(len(maxPeakIndex)): 
        frequencies.append(maxPeakIndex[i] * fs / N)
    return frequencies     
```

### Određivanje minimalnog broja uzoraka \( N \)

Da bismo odredili minimalni broj uzoraka \( N \) za koji prilikom izračunavanja diskretne Furijeove transformacije ne dolazi do curenja spektra (aliasing), koristimo odnos između frekvencije odabiranja $( f_s )$ i minimalne frekvencije u signalu $( f_{\text{max}} $).

Po formuli:

$$\
f = \frac{k \cdot f_s}{N}\
$$

gde je:
- $f_s$ frekvencija odabiranja,
- $f$ je maksimalna frekvencija signala,
- N je broj uzoraka,
- k je ceo broj koji u ovom slučaju mora biti pozitivan.

Za minimalno  N, uslov je da k bude ceo broj, da ne bi došlo do curenja spektra.
Kako su $f_s$ i f oba pozitivna odatle sledi da nam k mora biti pozitivno i to minimalno još pa zaključujemo da je k = 1.

kao što je rečeno, **minimalna frekvencija** u signalu je:

$$
f_1 = 1000 \, \text{Hz}
$$

jer je to najmanja frekvencija između:

$$
f_1 = 1000 \, \text{Hz}, \quad f_2 = 1300 \, \text{Hz}, \quad f_3 = 1320 \, \text{Hz}
$$

S obzirom na to, minimalni broj uzoraka \( N \) se računa kao:

\
$$N = \frac{f_s}{f_{\text{min}}} = \frac{11250}{1000} \= 11.250\$$


Pošto  N  mora biti ceo broj, uzimamo **najmanji ceo broj veći od 11.25**, što je **12**(Ovo se radi da bi sve tri frekvencije bile zadovoljene.

Dakle, **minimalni broj uzoraka je  N = 12**, kako bi spektar bio pravilno reprezentovan i kako ne bi došlo do curenja.

- Drugi nacin bi bio nacrtamo DTFT pa nakon DTFT nacrtamo DFT i nakon DFT i DTFT onda kad se to bude nacrtalo ako je DTFT odabiram u nulama DTFTa onda je spektar cist(ne curi) 
- Iteracijom(ja sam rucno mogla je i skripta) zakljucujemo da nema curenja spektra za **$N_{min} = 100.$**

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.fft as fft
import ipywidgets as widgets

# Funkcija za izračunavanje i prikazivanje FFT Amplitudne i Fazne karakteristike
def plotFFTAmplitudeAndPhase(N):
    # Odsecanje signala do N tačaka
    x_cut = x[:N]
    n = np.arange(N)
    fs=1000 # pretpostavka
    fk = np.arange(N)*fs / N
    w = np.ones(N)  # Pravougaoni prozor 
    xw = x_cut * w  # mnozenje prozorskom funkcijom

    # Računanje FFT-a
    X = fft.fft(xw)
    
    # Prikaz
    figure, ax = plt.subplots(2, 1, figsize=[6, 6])
    plt.subplots_adjust(bottom=0.25, left=0.15, hspace=1)

    # Amplitudska karakteristika
    Xa = np.abs(X)
    print(Xa)
    ax[0].stem(fk, Xa, basefmt=" ", use_line_collection=True)
    ax[0].set_title("Amplitudna karakteristika")
    ax[0].set_xlabel(r"$f_k$")
    ax[0].set_ylabel(r"$|X[k]|$")
    ax[0].grid(True)
    ax[0].set_xlim([0,fs/2])
    # Fazna karakteristika
    Xp = np.angle(X)
    Xp = Xp*(Xa > 1e-13) # ispravljena fazna karakteristika
    #print(Xa > 1e-13)
    ax[1].stem(fk, Xp, basefmt=" ", use_line_collection=True)
    ax[1].set_title("Fazna karakteristika")
    ax[1].set_xlabel(r"$f_k$")
    ax[1].set_ylabel(r"$\angle X[k]$")
    ax[1].grid(True)
    ax[1].set_xlim([0,fs/2])
    
    plt.show()

# widget prikaz iterativni 
plotFFTAmplitudeAndPhase(100)
```

```python
np,mpl,plt,signal = importEssentialLibs(True)
import scipy
import scipy.fft as fft
from scipy.io import wavfile

# Funkcija za računanje spektrograma

def dosSpectrogram(x, fs, window, noverlap, nfft, fmaxShow):
    N = nfft
    k = np.arange(nfft)
    f = k * fs / N #  računanje frekvencija na klasičan način 
    t = []
    numberOfWindows = (len(x) - len(window)) // (len(window) - noverlap) # broj prozora
    for i in range(numberOfWindows):
        t.append(i * (len(window) - noverlap) / fs)# računanje vremenskih trenutaka,smatram da je svaki početak vremena na početku prozora
    S = []
    for i in range(numberOfWindows):
        #Računanje matrice S,svaka vrsta je jedan vektor DFTA pa se zato posle transponuje, da svaka kolona bude vektor dfta i jos se castuje u array
        startOfWindow = (len(window) - noverlap) * i
        endOfWindow = startOfWindow + len(window)
        xw = x[startOfWindow:endOfWindow] * window
        w = scipy.signal.windows.hann(N,False)
        xw = xw * w
        X = fft.fft(xw, n=nfft)  
        S.append(X)
    S = np.array(S).T
    fmaxShow = min(fmaxShow, fs / 2)
    if np.isrealobj(x):
        fmaxIndex = int(nfft * fmaxShow // fs)
        f = f[0:fmaxIndex]
        S = S[:fmaxIndex, :]
    else:
        f = f[-fmaxShow:fmaxShow]
    return S, f, t
# Funkcija za prikaz spektrograma
def showSpectrogramRework(x, fs, fMaxShow=1600, Nwin=1024*4, noverlap=0):
    if noverlap == 0:
        noverlap = Nwin // 8  # postaviti default za noverlap
    
    window = signal.boxcar(Nwin)  # Boxcar prozor
    nfft = Nwin  # Broj tačaka za FFT

    # Izračunavanje spektrograma
    S, f, t = dosSpectrogram(x, fs, window, noverlap, nfft, fMaxShow)
    S = 20 * np.log10(np.maximum(abs(S), 1e-10))  # Logaritmuj
    # Indeks maksimalne frekvencije fMaxShow
    fMaxIndex = nfft * fMaxShow // fs

    # Napravi ok vidljivost grafika
    figure, axs = plt.subplots(figsize=[10, 3])
    plt.subplots_adjust(bottom=0.25, left=0.15)
    
    axs.matshow(S[:fMaxIndex, :], extent=[min(t), max(t), min(f), fMaxShow], origin='lower',cmap = "magma")
    #axs.pcolormesh(t, f[:fMaxIndex], S[:][:fMaxIndex], shading = 'nearest')
    forceAspect(axs, 3.5)
    
    # Obeležavanje osa
    axs.set_xlabel('$t$ [s]')
    axs.set_ylabel('$f$ [Hz]')
    axs.set_title("Spectrogram")
    
    plt.show()

# Učitavanje i prikaz zvučnih fajlova
fs1, x1 = wavfile.read("dz1_signali/nepoznati_broj_1.wav")
fs2, x2 = wavfile.read("dz1_signali/nepoznati_broj_2.wav")
fs3, x3 = wavfile.read("dz1_signali/nepoznati_broj_3.wav")

# Ispisivanje zvuka
import IPython
IPython.display.display(IPython.display.Audio(x1, rate=fs1))
IPython.display.display(IPython.display.Audio(x2, rate=fs2))
IPython.display.display(IPython.display.Audio(x3, rate=fs3))

time1 = np.arange(len(x1)) / fs1  

plt.figure(figsize=(10, 4))
plt.subplots_adjust(bottom=0.25,left=0.15)
plt.plot(time1, x1)  # Plotovanje signala,mora svaki 500-ti odbirak da bi se lepo video signal inače je ovako predebelo
plt.title('Signal u vremenskom domenu')
plt.xlabel('t  [s]')
plt.ylabel('Amplituda')
plt.grid(True)
plt.show()

time2 = np.arange(len(x2)) / fs2  

plt.figure(figsize=(10, 4))
plt.subplots_adjust(bottom=0.25,left=0.15)
plt.plot(time2, x2)  # Plotovanje signala
plt.title('Signal u vremenskom domenu')
plt.xlabel('t  [s]')
plt.ylabel('Amplituda')
plt.grid(True)
plt.show()

time3 = np.arange(len(x3)) / fs3  

plt.figure(figsize=(10, 4))
plt.subplots_adjust(bottom=0.25,left=0.15)
plt.plot(time3, x3)  # Plotovanje signala
plt.title('Signal u vremenskom domenu')
plt.xlabel('t  [s]')
plt.ylabel('Amplituda')
plt.grid(True)
plt.show()

# Prikaz spektrograma 
#showSpectrogramRework(x, fs, fMaxShow=1600, Nwin=1024*4, noverlap=0)
showSpectrogramRework(x1, fs1,2000,312,64)
showSpectrogramRework(x2, fs2,2000,384,64)
showSpectrogramRework(x3, fs3,2000,512,64)
```

```python
def fulfill_array(N):
    while not np.log2(N).is_integer():
        N += 1  # Povećaj dok ne postane stepen dvojke
    return np.zeros(N, dtype=complex)#vracam niz dopunjen nulama da mi to bude velicina niza pa cu posle da prepisem iz jednog u drugi

def fix_N(N):
    try:
        if N <= 0:
            raise ValueError("Value of N must be positive.")
        x = fulfill_array(N)
        return x
    except ValueError as ve:
        print("Error:", ve)
```

```python
import numpy as np

def fft_radix_2(x):
    N = len(x)
    x_fixed = fix_N(N)
    x_fixed[:N] = x
    X = np.zeros(N, dtype=complex)
    if(N == 1):
        return x;
    even = fft_radix_2(x[::2])
    odd = fft_radix_2(x[1::2])

    X = getRadixCoefficients(even, odd)
    return X

#arr_len mi je duzina nizova odd,even
# Xeven=[Xe0 Xe1], Xodd = [Xe2 Xe3] => X = FFT(x)
def getRadixCoefficients(even, odd):
    arr = np.zeros(2 * len(odd), dtype=complex)
    W = np.zeros(len(odd),dtype=complex)
    for k in range(len(odd)):
        N = 2 * len(odd)
        W[k] = np.exp(-1j *  2 * np.pi * k / N)
    for k in range(len(odd)):
        arr[k] = even[k] + W[k] * odd[k]
        arr[k + len(odd)] = even[k] - W[k] * odd[k]
    return arr

x = np.array([7, 2, 6, 3, 2, 2, 3, 9], dtype=complex)
X1 = fft_radix_2(x)
X2 = np.fft.fft(x)

print("Difference:", np.abs(X1 - X2))
```

## Zakljucak zasto se FFT rezultati razlikuju u mojoj implementaciji

Iako originalni signal ima 12 uzoraka, moja funkcija `fft_radix_2` automatski proširuje signal na 16 tačaka (zero-padding) da bi zadovoljila radix-2 FFT.

### Glavni razlozi razlike u rezultatima:

- **Broj tačaka (N = 16)**: `fft_radix_2(x)` koristi 16 tačaka, dok bibliotecka funkcija (`fft.fft(x)`) koristi 12 ako se eksplicitno ne navede padding.
- **Zero-padding menja spektar**: Dodavanjem nula signal postaje drugaciji — sto menja izgled spektra.
- **Curenje spektra**: Pošto signal nije periodičan unutar 16 tačaka, dolazi do curenja spektra, naročito bez prozorske funkcije.
